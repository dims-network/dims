## One series in detail

Pass the video duration if you know it — a series that does not span its
recording is the commonest cause of a dashboard that looks empty at one end.


```python
import glob, os
path = sorted(glob.glob(os.path.join(STUDY, 'assets/timeseries/*.csv')))[0]
signal_report(path)
```

```python
import matplotlib.pyplot as plt
t, v = load_series(path)
fig, ax = plt.subplots(2, 1, figsize=(11, 5), constrained_layout=True)
ax[0].plot(t, v, lw=.7); ax[0].set(title=os.path.basename(path), xlabel='time (s)')
import numpy as np
ax[1].plot(t[1:], np.diff(t), lw=.7)
ax[1].set(title='sampling interval — flat is good, spikes are gaps', xlabel='time (s)', ylabel='dt (s)')
plt.show()
```
